```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

```python
dd=pd.read_csv('C:/Users/saadv/OneDrive/Desktop/project1/diabetes.csv')
#dd['Outcome'].value_counts()
dd.groupby('Outcome').mean()
x = dd.drop(columns = 'Outcome')
y=dd['Outcome']
print(x)
print(y)
```

```python
scaler=StandardScaler()
scaler.fit(x)
standardizeddata=scaler.transform(x)
print(standardizeddata)
```

```python
x=standardizeddata
y=dd['Outcome']
print(x)
print(y)
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
print(x.shape,xtrain.shape,xtest.shape)
```

```python
classifer=svm.SVC(kernel='linear')
classifer.fit(xtrain,ytrain)
xtrainp=classifer.predict(xtrain)
xtrainaccuracry=accuracy_score(xtrainp,ytrain)
print(xtrainaccuracry)
```

```python

xtestp=classifer.predict(xtest)
xtesta=accuracy_score(xtestp,ytest)
print(xtesta)
```
